**VTAT and CTAT correlation With Customer Ratings**

---



```python
customerRatingData = df1[df1['Booking Status']=='Completed']

# creating bins
bins = range(0,int(customerRatingData['Avg CTAT'].max())+3,3)

# creating labels
lables = [f'{i} - {i+3}' for i in bins[:-1]]
#print(lables)
#print(customerRatingData)

# Bucket the Avg VTAT column
customerRatingData['CTAT_Group'] = pd.cut(
    customerRatingData['Avg CTAT'],
    bins=bins,
    labels=lables,
    right=False
)
#print(customerRatingData)

# Plot
plt.figure(figsize=(12,6))
sns.barplot(data = customerRatingData.groupby('CTAT_Group')['Customer Rating'].mean().reset_index(),
            x = 'CTAT_Group',
            y = 'Customer Rating',
            palette='viridis')
plt.title('Average Customer Rating vs Avg CTAT')
plt.xlabel('Avg CTAT Group (Minutes)')
plt.ylabel('Average Customer Rating')
plt.xticks(rotation=45)

plt.tight_layout()
plt.ylim(4.0, 4.5)
plt.show()
```

**ylim() function usage need,**
 - .describe() summary showed that the minimum customer rating in this entire dataset is 3.0 and the maximum is 5.0, with a massive concentration around 4.40. Therefore,
 - .mean() of each bucket, the averages are likely all hovering between 4.40 and 4.41. Because Matplotlib/Seaborn automatically scales the Y-axis from 0 to 4.5, a tiny change like 4.40 versus 4.41 looks completely flat to the naked eye.

**The Fix: Zoom in on the Y-Axis**

---
 - To reveal the hidden trend, we need to change the Y-axis limits so the graph doesn't start at 0. This will "stretch" the bars visually and show which time bucket actually drops in satisfaction.

**Vehicle Fleet Coverage**

---



```python
# 1. Group by Vehicle Type and calculate the raw metrics
fleet_summary = df1.groupby('Vehicle Type').agg(
    Total_Bookings=('Booking ID', 'count'),

    # Success Rate: % of rides where status is 'Completed'
    Success_Rate=('Booking Status', lambda x: (x == 'Completed').mean() * 100),

    # Average Distance per ride
    Avg_Distance=('Ride Distance', 'mean'),

    # Total Distance covered by that vehicle type
    Total_Distance=('Ride Distance', 'sum')
).reset_index()

# 2. Round the values to match professional reporting styles
fleet_summary['Success_Rate'] = fleet_summary['Success_Rate'].round(1).astype(str) + '%'
fleet_summary['Avg_Distance'] = fleet_summary['Avg_Distance'].round(2).astype(str) + ' km'

# 3. Format large numbers into millions (M) and thousands (K) for scannability
def format_large_numbers(val, is_distance=False):
    if val >= 1_000_000:
        return f"{val / 1_000_000:.2f}M"
    elif val >= 1_000:
        suffix = "K km" if is_distance else "K"
        return f"{val / 1_000:.0f}{suffix}"
    return str(val)

fleet_summary['Total_Bookings'] = fleet_summary['Total_Bookings'].apply(lambda x: format_large_numbers(x))
fleet_summary['Total_Distance'] = fleet_summary['Total_Distance'].apply(lambda x: format_large_numbers(x, is_distance=True))

# Display the final table
print(fleet_summary)
```
